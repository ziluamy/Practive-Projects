## Pandas

### Data Import
- read_csv
- iloc[]: only read the rows
- iloc[row, column]: locate to a specfic row and column
- df[column_name]: select a single column based on column name
- df[[name1,name2,...]]: the inner bracket serves as list. select multiple columns
- df[df["name1"] == "condition"]: filter the dataframe based on the column

```python
df = pd.read_csv('快递订单数据.csv')
df
```

```python
df.iloc[1:3,[0,3]] #filter by index to 
```

```python
df[(df['城市'] == '上海') & (df['重量(kg)'] > 2)]
```

### Group by & Agg (for multiple metrics)

```python
df.groupby('城市')['重量(kg)'].mean()
```

```python
df.groupby('城市')['重量(kg)'].agg(['mean','max','min']).rename(columns={'mean':'平均重量','max':'最大重量','min':'最小重量'})
```

### Merge and Join
- merge: df.merge(df2, on = 'column', how = 'left/right')
- join: you must set the index

```python
city_df = pd.read_csv('city_df.csv')
city_df.columns
```

```python
df_merged = df.merge(city_df, on='城市', how='left')
```

```python
df_merged = df_merged.dropna().drop_duplicates()
```

```python
df_merged
```
